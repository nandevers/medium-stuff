# Hierarchical Forecasting: Full Mathematical Breakdown

In this notebook, we break down the mathematical equations related to hierarchical forecasting, explaining why lower-level forecasting accuracy is crucial, especially in cases of high variability in price, cost, or time.

## 1. Revenue Variability Explained Mathematically

We’ll consider two products, **Product A** and **Product B**:

- Product A has a **high unit price** ($ P_A = 100 $) and **high variability** ($ \sigma_A = 30 $).
- Product B has a **low unit price** ($ P_B = 10 $) and **low variability** ($ \sigma_B = 5 $).

The **total revenue** ($ R $) is calculated as:

$$
R = P_A Q_A + P_B Q_B
$$

Where $ Q_A $ and $ Q_B $ represent the quantities sold of Product A and Product B, respectively.

## 2. Forecasting Scenario

If we forecast:
- $ \hat{Q}_A = 60 $ units of **Product A** (actual sales $ Q_A = 50 $),
- $ \hat{Q}_B = 400 $ units of **Product B** (actual sales $ Q_B = 500 $),

the **total forecast** may seem reasonable at the top level, but errors at the lower level can create significant revenue discrepancies.

## 3. Revenue Error Calculation

The forecast error in total revenue is:

$$
\Delta R = P_A (Q_A - \hat{Q}_A) + P_B (Q_B - \hat{Q}_B)
$$

For **Product A**, with a 10-unit forecast error:

$$
\Delta R_A = 100 \times (60 - 50) = 100 \times 10 = 1,000 \text{ dollars}
$$

For **Product B**, with a 100-unit forecast error:

$$
\Delta R_B = 10 \times (400 - 500) = 10 \times (-100) = -1,000 \text{ dollars}
$$

While the **total revenue** may seem balanced at the top level, errors in forecasting for **high-value, high-variability products** like Product A have a much larger financial impact, underscoring the importance of lower-level forecasting accuracy.

## 4. Variance in Revenue

Revenue variance ($ Var(R) $) represents the potential fluctuation in revenue due to errors in forecasting quantities and variability in prices.

For a product, revenue variance is calculated as:

$$
Var(R_A) = (E(Q_A))^2 \cdot Var(P_A) + (E(P_A))^2 \cdot Var(Q_A)
$$

For **Product A**:

$$
Var(R_A) = (50)^2 \cdot (30)^2 + (100)^2 \cdot Var(Q_A)
$$

For **Product B**:

$$
Var(R_B) = (500)^2 \cdot (5)^2 + (10)^2 \cdot Var(Q_B)
$$

This demonstrates that **Product A** has a much higher revenue variance due to its higher price and variability, further highlighting the need for accurate forecasting at lower levels.

## 5. MinT Forecast Reconciliation Formula

MinT (Minimum Trace) reconciliation is a method that adjusts forecasts across different levels of a hierarchy, minimizing forecast errors. The formula for MinT reconciliation is given by:

$$
\hat{y}_{\text{reconciled}} = S(S^T W^{-1} S)^{-1} S^T W^{-1} \hat{y}_{\text{base}}
$$

Where:
- $ S $ is the aggregation matrix linking lower-level forecasts to higher levels,
- $ W $ is the covariance matrix of forecast errors,
- $ \hat{y}_{\text{base}} $ is the base forecast at the lower levels.

## 6. Variance Minimization with MinT

MinT minimizes the overall forecast error variance across the hierarchy, ensuring that the reconciled forecasts are consistent with the base forecasts at both granular and aggregate levels. By minimizing the sum of squared errors, MinT provides more reliable forecasts for high-variability products or services.
